### Logical encoding of graphs with networkx

<p> The stabilizer tableau class can get a tableau from any networkx graph. The python file graphs has a function to create an encoded graph, a core graph in which every node is replaced by a given subgraph (logical graph) and the neighborhood of the nodes is adapted to the logical graph input node. </p>
The labels of the core and logical graph define the new labels in the encoded graph which gives a slight control to the node order in the encoded graph. 

```python
### Define the core and logical graphs
### The labels of the core and logical graphs define the labels in the resulting graph
### The encoded 0-node is labeled 0 and 1 from the logical labels and 1 from core has nodes 2 and 3
core_graph = nx.Graph([(0,1),(1,2),(2,3),(3,4),(4,0)])
logical_graph = nx.Graph([(0,1),(1,2)])
input_node = 1 #referring to an input node in the logical graph
###

### Call the encoding class
encoded_graph = encodedGraphs(core_graph,logical_graph,input_node).encoded_graph
```

```python
nx.draw(core_graph,with_labels = True)
plt.show()
nx.draw(logical_graph,with_labels = True)
```

```python
nx.draw(encoded_graph,with_labels=True,pos=nx.kamada_kawai_layout(encoded_graph))
```
