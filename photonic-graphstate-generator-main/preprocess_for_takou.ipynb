```python
from itertools import combinations


def create_G1():
    """
    Creates the bipartite graph G1:
    - Left side: 7 vertices labeled 1, 2, 3, 4, 5, 6, 7
    - Right side: 21 vertices labeled as sets {i,j,k,l,m} representing
      all 5-element subsets of the left side
    - Each right vertex is connected to exactly those 5 left vertices
    """
    G = nx.Graph()
    left_nodes = list(range(1, 8))
    G.add_nodes_from(left_nodes, bipartite=0)

    for subset in combinations(left_nodes, 5):
        node_label = "{" + ",".join(map(str, subset)) + "}"
        G.add_node(node_label, bipartite=1)
        for v in subset:
            G.add_edge(node_label, v)
    return G


G1 = create_G1()

# Compute bipartite layout
left_nodes = [n for n, d in G1.nodes(data=True) if d["bipartite"] == 0]
pos = nx.bipartite_layout(G1, left_nodes)

# Draw nodes and edges
nx.draw(G1, pos, with_labels=False, node_size=100,
        node_color=["skyblue" if d["bipartite"] == 0 else "lightgreen" for n, d in G1.nodes(data=True)])

# Separate labels for left and right nodes
labels_left = {n: n for n in left_nodes}
labels_right = {n: n for n in G1 if n not in left_nodes}

# Shift right-hand labels to the right
label_offset = 0.03  # adjust as needed
pos_labels_right = {n: (x + label_offset, y) for n, (x, y) in pos.items() if n in labels_right}

# Draw labels
nx.draw_networkx_labels(G1, pos, labels=labels_left, font_size=8)
nx.draw_networkx_labels(G1, pos_labels_right, labels=labels_right, font_size=6)

plt.show()
```

```python
g1_relabeled = nx.relabel_nodes(G1, {node: i for i, node in enumerate(G1.nodes())})

nx.draw(g1_relabeled, with_labels=True)

# Your adjacency matrix
adj_matrix = nx.adjacency_matrix(g1_relabeled).toarray()
# Save to .mat file
scipy.io.savemat(f'/Users/konstantinosrafailrevis/Documents/MATLAB/Optimal_Generation_Photonic_Graphs_Public/Adjacency_lclu_1.mat', {'Adj': adj_matrix})
```

```python
import networkx as nx
from itertools import combinations
import matplotlib.pyplot as plt

def create_G1_with_complete_left():
    """
    Creates a graph where:
    - Left side (1–7) form a complete graph K7
    - Right side: 21 vertices labeled by all 5-element subsets of {1,…,7}
    - Each right vertex connected to its corresponding 5 left vertices
    """
    G = nx.Graph()
    left_nodes = list(range(1, 8))
    G.add_nodes_from(left_nodes, bipartite=0)

    for subset in combinations(left_nodes, 5):
        label = "{" + ",".join(map(str, subset)) + "}"
        G.add_node(label, bipartite=1)
        for v in subset:
            G.add_edge(label, v)

    # Make left-hand nodes a complete graph (K7)
    for u, v in combinations(left_nodes, 2):
        G.add_edge(u, v)

    return G


# Create the graph
G2 = create_G1_with_complete_left()

# Identify left/right nodes
left_nodes = [n for n, d in G2.nodes(data=True) if d["bipartite"] == 0]
right_nodes = [n for n in G2 if n not in left_nodes]

# Layout: left-hand side as complete graph (K7)
pos_left = nx.circular_layout(left_nodes, scale=1.8)

# Shift the K7 to the left
for n in pos_left:
    pos_left[n] = (pos_left[n][0] - 2, pos_left[n][1])

# Arrange right-hand nodes vertically to the right
pos_right = {n: (2.5, i * 0.25) for i, n in enumerate(right_nodes)}

# Merge positions
pos = {**pos_left, **pos_right}

# Draw graph
nx.draw(
    G2,
    pos,
    with_labels=False,
    node_size=100,
    node_color=["skyblue" if n in left_nodes else "lightgreen" for n in G2.nodes()],
    edge_color=[
        "red" if (u in left_nodes and v in left_nodes) else "gray" for u, v in G2.edges()
    ],
    width=[
        0.8 if (u in left_nodes and v in left_nodes) else 0.5 for u, v in G2.edges()
    ]
)

# Draw labels (right ones slightly offset)
labels_left = {n: n for n in left_nodes}
labels_right = {n: n for n in right_nodes}
pos_labels_right = {n: (x + 0.2, y) for n, (x, y) in pos_right.items()}

nx.draw_networkx_labels(G2, pos, labels=labels_left, font_size=8)
nx.draw_networkx_labels(G2, pos_labels_right, labels=labels_right, font_size=6)

plt.axis("off")
plt.show()
```

```python
# Load a saved graph from the random graph database and set `input_graph`
import glob, os, re, pickle

def find_graph(node=None, p=None, idx=None, search_root='.'):
    """Search for a .gpickle file matching the given node, p and optional idx."""
    patterns = []
    if node is not None and p is not None:
        p_str = str(p)
        p_safe_dot = p_str
        p_safe_underscore = p_str.replace('.', '_')
        # Try multiple filename patterns used across the project
        patterns.extend([
            f"**/graph_db_node_{node}/**/*p{p_safe_dot}*.gpickle",
            f"**/graph_db_node_{node}/**/*p_{p_safe_underscore}*.gpickle",
            f"**/graph_db_node_{node}/**/*p{p_safe_underscore}*.gpickle",
        ])
    else:
        patterns.append("**/*.gpickle")

    for pat in patterns:
        files = glob.glob(os.path.join(search_root, pat), recursive=True)
        files = sorted(files)
        if files:
            if idx is not None:
                # Prefer files that encode an _idxNN pattern
                for f in files:
                    m = re.search(r'_idx(\d+)', f)
                    if m and int(m.group(1)) == int(idx):
                        return f
                # Fallback to positional index (1-based)
                if 0 < int(idx) <= len(files):
                    return files[int(idx)-1]
            return files[0]
    return None

# Example: change `node`, `p` and `idx` to select a different graph
node = 6
p = 0.65
idx = 1
path = find_graph(node=node, p=p, idx=idx, search_root='.')
if path is None:
    raise FileNotFoundError(f"No graph found for node={node}, p={p}, idx={idx}")
print('Loading graph from:', path)
try:
    input_graph = nx.read_gpickle(path)
except Exception:
    with open(path, 'rb') as f:
        input_graph = pickle.load(f)
print('Loaded graph with', input_graph.number_of_nodes(), 'nodes and', input_graph.number_of_edges(), 'edges')

# Compute adjacency matrix and save to .mat file
adj_matrix = nx.adjacency_matrix(input_graph).toarray()
scipy.io.savemat(f'/Users/konstantinosrafailrevis/Documents/MATLAB/Optimal_Generation_Photonic_Graphs_Public/big_run_data_base/test.mat', {'Adj': adj_matrix})
```

```python
# prepare the whole database of random graphs in matlab format
import os, glob, re, pickle
import scipy.io
import networkx as nx

OUT_ROOT = '/Users/konstantinosrafailrevis/Documents/MATLAB/Optimal_Generation_Photonic_Graphs_Public/big_run_data_base'

def parse_meta_from_filename(fname):
    basename = os.path.basename(fname)
    # Try to find p values like p0.65, p_0_65, p65 etc.
    m = re.search(r'p[_]?(\d+\.\d+|\d+)', basename)
    p = m.group(1) if m else 'unknown'
    # Try to extract a state name if present (best-effort).
    s = re.search(r'(rhg|tree|grid|cluster|lclu|random|state|graph)', basename, re.IGNORECASE)
    state = s.group(1).lower() if s else 'random'
    idx_m = re.search(r'_idx(\d+)', basename)
    idx = idx_m.group(1) if idx_m else None
    return p, state, idx

# Iterate nodes 6..30 and convert any graph files found to .mat adjacency matrices
for node in range(6, 31):
    print(f'Processing node {node}')
    pattern = f'**/graph_db_node_{node}/**/*'
    files = sorted(glob.glob(pattern, recursive=True))
    if not files:
        print(' - no files found for node', node)
        continue
    counter = 0
    for f in files:
        # Only process known graph serialization formats
        if not (f.endswith('.gpickle') or f.endswith('.pkl') or f.endswith('.pickle')):
            continue
        try:
            p, state, idx = parse_meta_from_filename(f)
            if idx is None:
                counter += 1
                idx = str(counter)
            out_dir = os.path.join(OUT_ROOT, f'node_{node}', f'p_{p}')
            os.makedirs(out_dir, exist_ok=True)
            print(' - loading', f)
            try:
                G = nx.read_gpickle(f)
            except Exception:
                with open(f, 'rb') as fh:
                    G = pickle.load(fh)
            adj = nx.adjacency_matrix(G).toarray()
            out_name = f'state_{state}_idx{idx}.mat'
            out_path = os.path.join(out_dir, out_name)
            scipy.io.savemat(out_path, {'Adj': adj})
            print('   saved', out_path)
        except Exception as e:
            print('   SKIP file', f, 'error:', e)

print('Done converting available graphs to .mat files under', OUT_ROOT)
```
