```python
tuples_for_rgh = [(1,1,1),(2,1,1),(3,1,1),(2,2,1),(3,2,1),(3,3,1),(2,2,2),(3,3,2),(3,3,3)]

rhg_dictionary = {}


for prms in tuples_for_rgh:
    rgh_graph = ogs.GraphState(shape='rhg', prms=prms)
    igraph_rgh = rgh_graph.graph
    networkx_rgh = igraph_rgh.to_networkx()
    print("Type of RHG lattice studied at the moment:", prms)
    opt_graph, circuit, stats = run_optimization(networkx_rgh)

    rhg_dictionary[prms] = {'graph': networkx_rgh, 'nodes': networkx_rgh.number_of_nodes(), 'edges': networkx_rgh.number_of_edges(), 'emitters': stats[0], 
                            'cnots': stats[1], 'gates': stats[2], 'emission_order': stats[3], 'local_complementations': stats[4], 
                            'opt_graph': opt_graph, 'circuit': circuit}

```

```python
import pandas as pd

# Define the parameters you want to include in the table
parameters = ['nodes', 'edges', 'emitters', 'cnots']

# Create an empty list to store the table rows
table_rows = []

# Iterate over the rhg_dictionary
for key, value in rhg_dictionary.items():
    # Extract the desired parameters from the value
    row_data = {param: value[param] for param in parameters}
    
    # Add the key to the row data
    row_data['Key'] = key
    
    # Add the row data to the list of table rows
    table_rows.append(row_data)

# Create a DataFrame from the table rows
table_df = pd.DataFrame(table_rows)

# Reorder columns
table_df = table_df[['Key'] + parameters]

# Rename columns for LaTeX output
table_df.columns = [
    "RHG lattice",
    r"$|V|$",
    r"$|E|$",
    r"$\# e$",
    r"$\# \mathrm{CNOT}$"
]

# Create the LaTeX table (without escaping LaTeX symbols)
latex_table = table_df.to_latex(index=False, escape=False)

# Replace tabular alignment specification
latex_table = latex_table.replace(r"\begin{tabular}{lrrrr}", r"\begin{tabular}{ccccc}")

# Print the modified LaTeX table
print(latex_table)
```
